```python
from Bio import SeqIO
```

```python
# 1. A
seqA = SeqIO.read('./a.fasta', format='fasta').seq
seqB = SeqIO.read('./b.fasta', format='fasta').seq

print('Length of sequence A:', len(seqA))
print('Length of sequence B:', len(seqB))
```

```python
# 1. B
print('Index of first ATC in sequence A:', seqA.index('ATC'))
print('Index of first ATC in sequence B:', seqB.index('ATC'))
```

```python
# 1. C
print(seqA[:15])
print(seqB[-35:])
seqC = seqA[:15] + seqB[-35:]
seqC = seqC[::-1]
```

```python
import matplotlib.pyplot as plt
```

```python
from collections import Counter

# 2. A
def freqCounter(seq, seqName):
  base_A = seq.count('A')
  base_C = seq.count('C')
  base_G = seq.count('G')
  base_T = seq.count('T')

  print('Sequence', seqName)
  print('Total A:', base_A)
  print('Total C:', base_C)
  print('Total G:', base_G)
  print('Total T:', base_T)
  print('=' * 15)

freqCounter(seqA, 'A')
freqCounter(seqB, 'B')
freqCounter(seqC, 'C')

counter_A = Counter(seqA)
counter_B = Counter(seqB)
counter_C = Counter(seqC)

plt.title('Sequence A')
plt.bar(counter_A.keys(), counter_A.values(), color="Red")
plt.show()
plt.title('Sequence B')
plt.bar(counter_B.keys(), counter_B.values())
plt.show()
plt.title('Sequence C')
plt.bar(counter_C.keys(), counter_C.values())
plt.show()
```

```python
# 2. B
def biCounter(seq, target_1, target_2):
  target_1_count = seq.count(target_1)
  target_2_count = seq.count(target_2)

  return (target_1_count + target_2_count) / len(seq) * 100

print(f'Sequence A:\nGC: {biCounter(seqA, 'G', 'C'):.2f}%\nAT: {biCounter(seqA, 'A', 'T'):.2f}%')
print('=' * 15)
print(f'Sequence B:\nGC: {biCounter(seqB, 'G', 'C'):.2f}%\nAT: {biCounter(seqA, 'A', 'T'):.2f}%')
print('=' * 15)
```

```python
# 2. C
from Bio.SeqUtils import MeltingTemp as mt

print('Melting temp for Sequence A:', mt.Tm_Wallace(seqA))
print('Melting temp for Sequence B:', mt.Tm_Wallace(seqB))
print('Melting temp for Sequence A:', mt.Tm_GC(seqA))
print('Melting temp for Sequence B:', mt.Tm_GC(seqB))
print('Melting temp for Sequence A:', mt.Tm_NN(seqA))
print('Melting temp for Sequence B:', mt.Tm_NN(seqB))
```

```python
# 2.D
from Bio.SeqUtils import molecular_weight as mw

print('Molecular weight for Sequence A:', mw(seqA), 'mol')
print('Molecular weight for Sequence B:', mw(seqB), 'mol')
print('Molecular weight for Sequence C:', mw(seqC), 'mol')
```

```python
# 3. A
seqA_mRNA = seqA.transcribe()
seqB_mRNA = seqB.transcribe()
seqC_mRNA = seqC.transcribe()

print('Sequence A transcribed:', seqA_mRNA)
print('Sequence B transcribed:', seqB_mRNA)
print('Sequence C transcribed:', seqC_mRNA)
```

```python
# 3. B
seqA_amino = seqA_mRNA.translate()
seqB_amino = seqB_mRNA.translate()
seqC_amino = seqC_mRNA.translate()

print(seqA_mRNA)
print(seqB_mRNA)
print(seqC_mRNA)
```

```python
# 4. A
from Bio import pairwise2 as pw

localxx = pw.align.localms(seqA, 'ATCGTACGGTACCGTAA', 2, -1, -2, -0.5)
print(pw.format_alignment(*localxx[0]))
```

```python
globalxx = pw.align.globalms(seqA, 'GGTACCGTAGCTAGGCTAATCG', 2, -1, -2, -0.5)
print(pw.format_alignment(*globalxx[0]))
```
